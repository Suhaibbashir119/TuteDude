Valid Identifiers

```python
valid_identifier1 = "_name"
valid_identifier2 = "Salary"
valid_identifier3 = "totalRev2026"
```

Invalid Identifiers

```python
invalid_identifier1 = "2name" #Invalid : cannot start with the number
invalid_identifier2 = "class" # Invalid: cannot use keyword as identifiers
```

```python
print("Valid Identifiers : " , valid_identifier1, valid_identifier2, valid_identifier3)
print("Invalid Identifiers (may cause errors if used):" , invalid_identifier1, invalid_identifier2)
```

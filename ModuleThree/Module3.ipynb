# Python Operators

#### Arithmetic Operators:

##### 1: Addition(+):  

        Dsecription: Used to add two numbers.  
        Note: Commonly used in billing, financial computations.

```python
total = 150 + 200
print("Total : ",total)
```

```python
a = 10
b = 30
total = a + b
print("Total : ",total)
```

#### 2: Subtraction(-):  

        Description: Used to Subtract one number from another.    
        Note: Helpful for calculating differences like discounts or balances.

```python
balance = 1000
balance = balance + 500 #balance is now 1500
print("Balance : ",balance)
```

#### 3: Multiplication(*):  

        Description: Used to multiply two numbers.  
        Note: Useful for finding total cost = unit_price x quantity

```python
price = 40
qty = 5
total_cost = price * qty
print("Total Cost: ",total_cost)
```

#### 4: Division(/):  

        Description: Divides a numerator by the denominator.  
        Note: Returns a float value even if result is whole(in python). 

```python
avg = 100/5 
print("Average: ", avg)
```

#### 5: Floor Division(//) :   

        Description: divides and returns the intiger part only.  
        Note: Used where whole units are important eg: box count. 

```python
boxes = 17 // 5
print("Full boxes: ", boxes)
```

#### 6: Modulus(%):  

        Description: Returns a remainder of a division.  
        Note: Often used in even/odd checks or cycle repititions.

```python
remainder = 19 % 6
print("Modulus: ", remainder)
```

#### 6: Exponentiation(**):  

        Description: Raises one number to the power of another.  
        Note: Used in financial projections or mathematical computations.

```python
compound = 4**5
print("Exponentiation: ", compound)
```

# Case Study: Retail Billing (Kirana Store)
### Problem Statement:
A kirana store needs a billing system that calculates the total price for each item(price x quantity), applies a flat discount , and shows a final bill.
We'll use Arithmetic Operators to implement this: 

```python
# item Details
item1_price, item1_qty = 40 , 5
item2_price, item2_qty = 30, 6

#Calculate Totals
item1_total= item1_price * item1_qty
item2_total = item2_price * item2_qty

#subTotal
subtotal = item1_total + item2_total

# Discount in rupees 
discount = 30
final_total = subtotal - discount

#print all now
print("Subtotal: ",subtotal)
print("Discount: ",discount)
print("Final Total Bill: ", final_total)
```

# Assignment Operators

These are used to assign values to variable and modify them using various arthematic operations.

#### 1: Assign (=)
        Description: Assigns the value on the right to the variable on right.
        Note: Most basic form of assignment.

```python
stock = 100
print("Initial Stock: ", stock)
```

#### 2: Add and Assign (+=)
        Description: Adds and assigns the result to the variable.
        Note: Useful for cumulative totals.

```python
stock = 100
stock += 20
print("Updated Stock: ", stock)
```

#### 3: Subtract and Assign(-=)
        Description: Subtracts and assigns result to the variable.
        Note: Used in decrementing inventory.

```python
stock = 120
stock -= 30
print("Remaining Stock after Sale: ", stock)
```

#### 4: Multiply and Assign (*=)
        Description: Multiplies and assigns result to the variable.
        Note: Used in progressive increase.

```python
stock = 120
stock *= 2
print("Stock after Restocking: ", stock)
```

#### 5: Divide and Assign(/=)
        Description: Divides and assigns result to the variable.
        Note: Used in averaging and scaling.

```python
stock = 100
stock /= 2
print("Stock after Division: ", stock)
```

#### 6: Modulus and Assign(%=)
        Description: Applies modulus and assigns result.
        Note: Useful for wrapping or cycles.

```python
count = 10
count %= 3
print("Count after Modulus: ", count)
```

# Bitwise Operators
These operators work on bits and perform bit-by-bit operations.

#### 1: AND ( & )
        Description: Performs bitwise AND.
        Note: Used in IoT for permission flags or status registers.

```python
#Imaagine we have two simple numbers

a = 5
b = 10

#Bitwise AND compares each bit of the first operand to the corresponding bit of the second operand. 
# If both bits are 1, the corresponding result bit is set to 1. Otherwise, it is set to 0.

result_and = a & b
print("Bitwise AND of ", a, "and", b, "is: ", result_and)
```

#### 2: OR( | )
        Description: Performs Bitwise OR
        Note: Combines flags or activates multiple bits.

```python
or__result = a | b
print("Bitwise OR of ", a, "and", b, "is: ", or__result)
```

### Identity and Membership Operators
Used to compare memory location and collection membership.

#### 1: Identity Operator : is
        Description: Checks if two variables point to the same object.
        Note: Used for Singleton checks like None. 

```python
a = None
print("Is none :", a is None)
```

#### 2: Identity Operator : is not
        Description: Checks if they do not point to the same object.
        Note: Validates non-identity.

```python
a= 10
print("Is not None: ", a is not None)
```

#### 3: Membership Operetor: in
        Description: Checks if a value is in a sequence(list, Tuple etc).
        Note: Common in filtering and access control.

```python
skills=["Python", "Java", "C++", "JavaScript"]
print("Skills List: ", skills)
print("Java in Skills List: ", "Java" in skills)
print("Python in Skills List: ", "Rust" in skills)
```

#### 4: Membership Operator : not in
        Description: Checks if a value is not in a sequence(List, Tuple etc)
        Note: Used to Exclude Values.

```python
print("Java not in Skills List: ", "Java" not in skills)
print("Rust not in Skills List: ", "Rust" not in skills)
```

# Case Study: HR Skill Matching System
    Problem Statement:
An HR system needs to check if a candidate has required skills using Identity and Membership Operators?    

```python
candidate ={
    "name": "Alice",
    "age": 30,
    "skills": ["Python", "Java", "C++", "JavaScript", "Communication"]
}

# Check if candidate has the skill "Python" and "Communication"


#Case 1: Candidate has both skills
if "Python" in candidate["skills"] and "Communication" in candidate["skills"]:
    print("Candidate has both Python and Communication skills.")
else:
    print("Candidate does not have both Python and Communication skills.")
    

#Case 2: Candidate has at least one of the skills
if "Python" in candidate["skills"] or "Communication" in candidate["skills"]:
    print("Candidate has at least one of the skills: Python or Communication.")
else:
    print("Candidate does not have either Python or Communication skills.") 
```

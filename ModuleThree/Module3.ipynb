# Python Operators

#### Arithmetic Operators:

##### 1: Addition(+):  

        Dsecription: Used to add two numbers.  
        Note: Commonly used in billing, financial computations.

```python
total = 150 + 200
print("Total : ",total)
```

```python
a = 10
b = 30
total = a + b
print("Total : ",total)
```

#### 2: Subtraction(-):  

        Description: Used to Subtract one number from another.    
        Note: Helpful for calculating differences like discounts or balances.

```python
balance = 1000
balance = balance + 500 #balance is now 1500
print("Balance : ",balance)
```

#### 3: Multiplication(*):  

        Description: Used to multiply two numbers.  
        Note: Useful for finding total cost = unit_price x quantity

```python
price = 40
qty = 5
total_cost = price * qty
print("Total Cost: ",total_cost)
```

#### 4: Division(/):  

        Description: Divides a numerator by the denominator.  
        Note: Returns a float value even if result is whole(in python). 

```python
avg = 100/5 
print("Average: ", avg)
```

#### 5: Floor Division(//) :   

        Description: divides and returns the intiger part only.  
        Note: Used where whole units are important eg: box count. 

```python
boxes = 17 // 5
print("Full boxes: ", boxes)
```

#### 6: Modulus(%):  

        Description: Returns a remainder of a division.  
        Note: Often used in even/odd checks or cycle repititions.

```python
remainder = 19 % 6
print("Modulus: ", remainder)
```

#### 6: Exponentiation(**):  

        Description: Raises one number to the power of another.  
        Note: Used in financial projections or mathematical computations.

```python
compound = 4**5
print("Exponentiation: ", compound)
```

# Case Study: Retail Billing (Kirana Store)
### Problem Statement:
A kirana store needs a billing system that calculates the total price for each item(price x quantity), applies a flat discount , and shows a final bill.
We'll use Arithmetic Operators to implement this: 

```python
# item Details
item1_price, item1_qty = 40 , 5
item2_price, item2_qty = 30, 6

#Calculate Totals
item1_total= item1_price * item1_qty
item2_total = item2_price * item2_qty

#subTotal
subtotal = item1_total + item2_total

# Discount in rupees 
discount = 30
final_total = subtotal - discount

#print all now
print("Subtotal: ",subtotal)
print("Discount: ",discount)
print("Final Total Bill: ", final_total)
```

# Comparision operators

These operators are used to compare values. athey return Boolean values: True and false...

#### 1: Equal To(==)  

        Description: Checks if two values are equal.  
        Note: Commonly used in Conditionalls and Validations.

```python
marks = 75
print("Passed == 75: ", marks == 75)
```

```python
marks = 76
print("Passed == 75: ", marks == 75)
```

#### 2: Not Equal (!=)

        Description: Checks if two values are not equal.  
        Note: Used for execution logic.

```python
print('Absent != 0: ', marks != 0)
```

#### 3: Greater Than(>)

        Description: Checks if the left value is greater than right value.  
        Note: Useful in grading, threshholds.

```python
print("Grade A: ", marks > 70)
```

#### 4: Less Than (<)

        Description: Checks if left value is less than the right value.  
        Note: Used in limit Checks and Filtering.

```python
marks = 75
print("Needs Improvement: ", marks < 70)
```

```python
marks = 65
print("Needs Improvement: ", marks < 70)
```

#### 5: Greater Than or Equal To (>=)
        Description: Checks if the left value is greater or equal.  
        Note: Used in Pass/Fail conditions.

```python
print('First Class: ', marks >= 65)
```

#### 6: Less Than or Equal TO (<=)  

        Description: Checks if left value is less or equal than right value.  
        Note: Used in eligiblity logic.

```python
print('Not Eligible: ', marks)
```

# Case Study: Education Result Evaluation  
### Problem Statement: 
Create a program that executes student results using marks and assigns performance categories like 'Distinction' , 'First Class' , 'Fail'. We will use the **Comparision Operators** here.

```python
marks = 80 # Change value here for checking other elif statements

if marks >= 75:
    category = "Distinction"
elif marks >= 60:
    category = "First Class"
elif marks >= 40:
    category = "Fail Class"
else:
    category = "Fail"
    

print("Student Marks: ", marks)
print("Category: ", category)    
```

# Logical Operators

These operators are used to combine conditional statements.  

#### 1: Logical AND (and)  
        Description: Returns True if both conditions are true.  
        Note: Used in multi-condition filtering.

```python
bp = 135
sugar = 160
print("Critical Condition: ", bp > 130 or sugar > 150)
```

#### 2: Logical OR (or)

        Description: Returns True if atleast one condition is true.  
        Note: Used in decision Trees.

```python
print("Requires Immediate Attention: ", bp > 130 and sugar > 150)
```

#### 3: Logical NOT (not)

        Description: Inverts the Boolean value.  
        Note: Used to toggle logic.

```python
healthy = False
print("Not Healthy: ", not healthy)
```

### Case Study: Healthcare Screening
    problem Statement:
In a screening program, check if a patient is at health risk based on BP and Sugar levels using logical operators.


```python
bp = 150
sugar = 160

if bp > 130 and sugar > 150:
    status = "Requires Immediate Attention"
elif bp > 120 or sugar > 140:
    status = "Needs Attention"
else:
    status = "Healthy"
    
print("Blood Pressure: ", bp , "| Sugar Level: ", sugar)
print("Health Status: ", status)    
    
```

Q: Mention Reasons for invalid identifiers from following list:  
1: 2var  
2: my-name  
3: class  
4: first name   
5: price$  

Answer:  
1: 2var--> identifier cannot start with number   
2: my-name--> we can only use underscore(_) in identifiers   
3: class--> class is a keyword , so it cannot be used as identifier  
4: first name--> spaces are not allowed in defining the identifier  
5: price$--> no special character($) is allowed to use in defining identifier
